```python
def sm2acc(sm, n, a=1024):
    sm = np.asarray(sm)
    a = np.random.default_rng().random(sm.shape + (a, n - 1))
    return (a.max(-1) / a.sum(-1) * (1 - sm[..., None]) < sm[..., None]).mean(-1)


def sm2accx(n, Y=0.5):
    l, x, r = 1 / n, 2 / n, .5
    y = sm2acc(x, n)
    while abs(y - Y) > 1e-3:
        if y < Y:
            l = x
            x = (x + r) / 2
        else:
            r = x
            x = (l + x) / 2
        y = sm2acc(x, n)
    return x
```

```python
plt.loglog([sm2accx(i) for i in range(4, 400)], label=".5")
plt.loglog([sm2accx(i, 0.9) for i in range(4, 400)], label=".9")
plt.loglog([sm2accx(i, 0.1) for i in range(4, 400)], label=".1")
plt.legend()
```
