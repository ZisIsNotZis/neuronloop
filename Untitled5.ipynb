```python
from torch.nn import *
import torch


def Conv1dFwd(x, w):
    return torch.fft.ifft(torch.fft.fft(x)[:, None] * torch.fft.fft(w.flip(2), x.size(2))).real.sum(2).roll(-1, 2) + s.bias[:, None]


def Conv1dInv(x, w):
    return torch.fft.ifft(torch.fft.fft((b - s.bias[:, None]).roll(1, 2) / w.size(1))[:, :, None] / torch.fft.fft(w.flip(2), b.size(2))).real.mean(1)


def ConvChannelwise1dFwd(x, w):
    return torch.fft.ifft(torch.fft.fft(x) * torch.fft.fft(w[None, :, 0].flip(2), x.size(2))).real.roll(-1, 2)


def ConvChannelwise1dInv(x, w):
    return torch.fft.ifft(torch.fft.fft(x.roll(1, 2)) / torch.fft.fft(w[None, :, 0].flip(2), x.size(2))).real


def ConvPointwise1dFwd(x, w):
    return w[..., 0] @ x


def ConvPointwise1dInv(x, w):
    return torch.linalg.lstsq(w[None, ..., 0], x)[0]


def LinearInv(x, w):
    return torch.linalg.lstsq(w.T, y.T)[0].T
```

Letting $\mathbb{K}$ be $\mathbb{R}$ or :math:`\mathbb{C}`,
the **least squares problem** for a linear system :math:`AX = B` with
:math:`A \in \mathbb{K}^{m \times n}, B \in \mathbb{K}^{m \times k}` is defined as


$$\min_{X \in \mathbb{K}^{n \times k}} \|AX - B\|_F$$

where :math:`\|-\|_F` denotes the Frobenius norm.
