```python
from nn import *

A = np.lib.stride_tricks.sliding_window_view(np.memmap("../datbrick/afqmc+iflytek+inews.H", "h", "r"), 2048)
unvoc = np.array(open("../datbrick/afqmc+iflytek+inews.txt", encoding="utf8").read().split("\n"), "U1")
```

```python
def Conv(k, r=1):
    return Add(
        Seq(
            Conv1d(k, k, r * 2 + 1, padding=r, bias=False),
            BatchNorm1d(k, track_running_stats=False),
            SiLU(True),
        ),
        Conv1d(k, k, 1, bias=False),
    )


def Attn(k, h=1):
    return Res(LiftT(lambda x, _=torch.nn.MultiheadAttention(k, h, batch_first=True): _(x, x, x)[0]))


def Pool(k, *_):
    return Res(
        Conv1d(k, k, 1, bias=False),
        MaxPool1d(2),
        Seq(*_),
        ConvTranspose1d(k, k, 3, 2, 1, 1, bias=False),
    )


k = 256
m = Seq(
    RandHot(len(unvoc), p=np.log(1 - 0.75 / len(unvoc)) / np.log(0.5)),
    Peek(F(globals().__setitem__, "X", F.X)),
    F(torch.matmul, Embedding(len(unvoc), k).weight),
    T,
    Res(
        Conv(k, 4),
        Conv(k),
        Pool(
            k,
            Conv(k),
            Conv(k),
            Pool(
                k,
                Conv(k),
                Conv(k),
                Pool(
                    k,
                    Conv(k),
                    Conv(k),
                    Pool(
                        k,
                        Conv(k),
                        Conv(k),
                        Pool(
                            k,
                            Conv(k),
                            Conv(k),
                            Pool(
                                k,
                                Conv(k),
                                Conv(k),
                                Pool(
                                    k,
                                    Conv(k),
                                    Conv(k),
                                    Pool(
                                        k,
                                        Conv(k),
                                    ),
                                    Conv(k),
                                    Conv(k),
                                ),
                                Conv(k),
                                Conv(k),
                            ),
                            Conv(k),
                            Conv(k),
                        ),
                        Conv(k),
                        Conv(k),
                    ),
                    Conv(k),
                    Conv(k),
                ),
                Conv(k),
                Conv(k),
            ),
            Conv(k),
            Conv(k),
        ),
        Conv(k),
        Conv(k),
    ),
    Conv1d(k, len(unvoc), 1),
    T,
).cuda()
o = torch.optim.Adam(m.parameters())
O = torch.optim.lr_scheduler.OneCycleLR(o, 0.0003, 100000, pct_start=0.01, div_factor=9, final_div_factor=99)
# s = torch.cuda.amp.GradScaler()
```

```python
torch.save((m,o,O), '256.pk', pickle_module=__import__('dill'))
```

```python
from tqdm.notebook import tqdm


def show(o, a, b, **_):
    color = np.full(len(o), "\33[0m", "U5")
    color[(o == a) & (o != b)] = "\33[31m"
    color[(o != a) & (o == b)] = "\33[32m"
    color[(o != a) & (o != b)] = "\33[33m"
    print("".join(np.char.add(color, unvoc[o])), end="\33[0m\n", **_)


with open("log.txt", "w", encoding="utf8") as f:
    for i in tqdm(range(O.last_epoch, O.total_steps)):
        # with torch.cuda.amp.autocast():
        x = torch.tensor(A[np.random.randint(0, len(A), 12)], dtype=torch.long, device="cuda")
        y = m(x)
        l = functional.cross_entropy(y.transpose(1, 2), x)
        # s.scale(l).backward()
        # s.step(o)
        # s.update()
        l.backward()
        o.step()
        o.zero_grad()
        O.step()
        if not (i + 1) % 20:
            print(end=f"{l:.3f} ", file=f)
            if not (i + 1) % 200:
                print((y.argmax(2) == x).sum() / x.numel(), file=f)
                show(x[0].cpu(), X[0].argmax(1).cpu(), y[0].argmax(1).cpu(), file=f)
                f.flush()
```
