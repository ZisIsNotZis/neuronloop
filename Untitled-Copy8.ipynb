```python
from nn import *
from ipywidgets import Output
from tqdm.notebook import tqdm
import matplotlib.pyplot as plt


class lmap(map):
    __slots__ = ("_",)

    def __init__(s, f, _):
        s._ = len(_)

    def __len__(s):
        return s._


def train(x, m, f, **_):
    o = torch.optim.Adam(m.parameters(), eps=1e-9)
    O = torch.optim.lr_scheduler.OneCycleLR(o, total_steps=len(x), **_)
    for x, t in x:
        y = m(x)
        l, a = f(y, t)
        l.backward()
        o.step()
        o.zero_grad()
        O.step()
        yield a


def log(x, res=100, ref=1000):
    X = [0]
    o = Output()
    display(o)
    for i, x in enumerate(x, 1):
        X[-1] += x
        if not i % res:
            X[-1:] = X[-1][None] / res, 0
            if not i % ref:
                X = [torch.cat([X.cpu() for X in X[:-1]], 0), 0]
                plt.plot(X[0])
                with o:
                    o.clear_output(True)
                    plt.show()
    return X[0]


def squnflatten(x, dim=-1):
    return x.unflatten(dim, [-1, int(x.size(dim) ** 0.5)])


def Bitention(i, o, head=1, group=1):
    i = i // group + 1
    o //= group
    return Seq(
        Unflatten(-2, [group, i - 1]),
        F(functional.pad, [0, 0, 0, 1], value=1),
        Einsum(
            "...ahbc...kc,...jb,...ohjk->...oa",
            Seq(
                Einsum(
                    "...hijk,...kc,...jb,...ia->...ahbc",
                    Param(group, head, i, i, i),
                    Identity(),
                    Identity(),
                    Identity(),
                ),
                Flatten(-2),
                Softmax(-1),
                F(squnflatten),
            ),
            Identity(),
            Identity(),
            Param(group, o, head, i, i),
        ),
        Flatten(-3, -2),
    )


def trispace(x, dim=-1, kdim=-2, cache={}):
    if (_ := cache.get(x.size(dim))) is None:
        _ = torch.linspace(0, 2 * torch.pi, x.size(dim), device=x.device)
        _ = cache[x.size(dim)] = torch.stack([torch.sin(_), torch.cos(_)])
    return _[:, *[None] * (dim - kdim - 1), :, *[None] * (-dim - 1)]


def Attention(i, o, r=0, head=1, group=1):
    i //= group
    o //= group
    return Seq(  # n(gi)a
        Dot(
            Seq(
                Dot(
                    Seq(
                        Conv1d(group * i, group * head * i, 2 * r + 1, padding=r, groups=group),  # n(ghi)a
                        Unflatten(-2, [group, head, i]),
                        T,
                    ),  # nghai
                    Unflatten(-2, [group, 1, i]),  # ng1ib
                ),  # nghab
                Softmax(-1),
            ),  # nghab
            Seq(
                Unflatten(-2, [group, 1, i]),
                T,
            ),  # ng1bi
        ),  # nghai
        T,
        Flatten(-4, -2),  # n(ghi)a
        Conv1d(group * head * i, group * o, 2 * r + 1, padding=r, groups=group),  # n(go)a
    )  # n(go)a


def ConvAttention(i, h, R=0, r=0, head=0, pos=1, bn=1, act=SiLU(True), res=1, group=1, bias=False):
    return Add(
        Seq(
            Add(
                head and Attention(i, h, r, head=head, group=group),
                Conv1d(i, h, 2 * R + 1, padding=R, groups=group, bias=not bn),
                pos
                and Seq(
                    F(trispace),
                    Conv1d(2, h, 1, bias=False),
                ),
            ),
            bn and BatchNorm1d(h, 1e-9, track_running_stats=False),
            act,
            Conv1d(h, h, 2 * r + 1, padding=r, groups=group, bias=bias),
        ),
        res and Conv1d(i, h, 2 * r + 1, padding=r, groups=group, bias=False),
    )


def Attention2d(i, o, r=0, head=1, group=1):
    i //= group
    o //= group
    return Seq(  # n(gi)ab
        Dot(
            Seq(
                Dot(
                    Seq(
                        Conv2d(group * i, group * head * i, 2 * r + 1, padding=r, groups=group),  # n(ghi)ab
                        Unflatten(-3, [group, head, i]),
                        F(torch.permute, [0, 1, 2, 4, 5, 3]),
                    ),  # nghabi
                    Seq(
                        Unflatten(-3, [group, 1, i]),
                        F(torch.transpose, -3, -2),
                    ),  # ng1aic
                ),  # nghabc
                Softmax(-1),
            ),  # nghabc
            Seq(
                Unflatten(-3, [group, 1, i]),
                F(torch.permute, [0, 1, 2, 4, 5, 3]),
            ),  # ng1aci
        ),  # nghabi
        F(torch.permute, [0, 1, 2, 5, 3, 4]),
        Flatten(-5, -3),  # n(ghi)ab
        Conv2d(group * head * i, group * o, 2 * r + 1, padding=r, groups=group),  # n(go)ab
    )  # n(go)ab


def ConvAttention2d(i, h, R=0, r=0, yhead=0, xhead=0, ypos=1, xpos=1, bn=1, act=SiLU(True), res=1, group=1, bias=False):
    return Add(
        Seq(
            Add(
                yhead and Attention2d(i, h, r, head=yhead, group=group),
                xhead and Seq(T, Attention2d(i, h, r, head=xhead, group=group), T),
                Conv2d(i, h, 2 * R + 1, padding=R, groups=group, bias=not bn),
                ypos
                and Seq(
                    F(trispace, -1, -3),
                    Conv2d(2, h, 1, bias=False),
                ),
                xpos
                and Seq(
                    F(trispace, -2, -3),
                    Conv2d(2, h, 1, bias=False),
                ),
            ),
            bn and BatchNorm2d(h, 1e-9, track_running_stats=False),
            act,
            Conv2d(h, h, 2 * r + 1, padding=r, groups=group, bias=bias),
        ),
        res and Conv2d(i, h, 2 * r + 1, padding=r, groups=group, bias=False),
    )
```

```python
from nn import *
from ipywidgets import Output
from tqdm.notebook import tqdm
import matplotlib.pyplot as plt

def train(x, m, o, O, f, **_):
    for x, t in x:
        y = m(x)
        l, a = f(y, t)
        l.backward()
        o.step()
        o.zero_grad()
        O.step()
        yield a


def log(x, res=100, ref=1000):
    X = [0]
    o = Output()
    display(o)
    try:
        for i, x in enumerate(x, 1):
            X[-1] += x
            if not i % res:
                X[-1:] = X[-1][None] / res, 0
                if not i % ref:
                    X = [torch.cat([X.cpu() for X in X[:-1]], 0), 0]
                    plt.plot(X[0])
                    with o:
                        o.clear_output(True)
                        plt.show()
    except KeyboardInterrupt:
        pass
    finally:
        X = torch.cat([X.cpu() for X in X[:-1]], 0)
        plt.plot(X)
        o.clear_output()
        plt.show(X)
    return X


def squnflatten(x, dim=-1):
    return x.unflatten(dim, [-1, int(x.size(dim) ** 0.5)])


def Bitention(i, o, head=1, group=1):
    i = i // group + 1
    o //= group
    return Seq(
        Unflatten(-2, [group, i - 1]),
        F(functional.pad, [0, 0, 0, 1], value=1),
        Einsum(
            "...ahbc...kc,...jb,...ohjk->...oa",
            Seq(
                Einsum(
                    "...hijk,...kc,...jb,...ia->...ahbc",
                    Param(group, head, i, i, i),
                    Identity(),
                    Identity(),
                    Identity(),
                ),
                Flatten(-2),
                Softmax(-1),
                F(squnflatten),
            ),
            Identity(),
            Identity(),
            Param(group, o, head, i, i),
        ),
        Flatten(-3, -2),
    )


def trispace(x, dim=-1, kdim=-2, cache={}):
    if (_ := cache.get(x.size(dim))) is None:
        _ = torch.linspace(0, 2 * torch.pi, x.size(dim), device=x.device)
        _ = cache[x.size(dim)] = torch.stack([torch.sin(_), torch.cos(_)])
    return _[(slice(None), [None] * (dim - kdim - 1), slice(None), *[None] * (-dim - 1))]


def Attention(i, o, r=0, head=1, group=1):
    i //= group
    o //= group
    return Seq(  # n(gi)a
        Dot(
            Seq(
                Dot(
                    Seq(
                        Conv1d(group * i, group * head * i, 2 * r + 1, padding=r, groups=group),  # n(ghi)a
                        Unflatten(-2, [group, head, i]),
                        T,
                    ),  # nghai
                    Unflatten(-2, [group, 1, i]),  # ng1ib
                ),  # nghab
                Softmax(-1),
            ),  # nghab
            Seq(
                Unflatten(-2, [group, 1, i]),
                T,
            ),  # ng1bi
        ),  # nghai
        T,
        Flatten(-4, -2),  # n(ghi)a
        Conv1d(group * head * i, group * o, 2 * r + 1, padding=r, groups=group),  # n(go)a
    )  # n(go)a


def ConvAttention(i, h, R=0, r=0, head=0, pos=1, bn=1, act=SiLU(True), res=1, group=1, bias=False):
    return Add(
        Seq(
            Add(
                head and Attention(i, h, r, head=head, group=group),
                Conv1d(i, h, 2 * R + 1, padding=R, groups=group, bias=not bn),
                pos
                and Seq(
                    F(trispace),
                    Conv1d(2, h, 1, bias=False),
                ),
            ),
            bn and BatchNorm1d(h, 1e-9, track_running_stats=False),
            act,
            Conv1d(h, h, 2 * r + 1, padding=r, groups=group, bias=bias),
        ),
        res and Conv1d(i, h, 2 * r + 1, padding=r, groups=group, bias=False),
    )


def Attention2d(i, o, r=0, head=1, group=1):
    i //= group
    o //= group
    return Seq(  # n(gi)ab
        Dot(
            Seq(
                Dot(
                    Seq(
                        Conv2d(group * i, group * head * i, 2 * r + 1, padding=r, groups=group),  # n(ghi)ab
                        Unflatten(-3, [group, head, i]),
                        F(torch.permute, [0, 1, 2, 4, 5, 3]),
                    ),  # nghabi
                    Seq(
                        Unflatten(-3, [group, 1, i]),
                        F(torch.transpose, -3, -2),
                    ),  # ng1aic
                ),  # nghabc
                Softmax(-1),
            ),  # nghabc
            Seq(
                Unflatten(-3, [group, 1, i]),
                F(torch.permute, [0, 1, 2, 4, 5, 3]),
            ),  # ng1aci
        ),  # nghabi
        F(torch.permute, [0, 1, 2, 5, 3, 4]),
        Flatten(-5, -3),  # n(ghi)ab
        Conv2d(group * head * i, group * o, 2 * r + 1, padding=r, groups=group),  # n(go)ab
    )  # n(go)ab


def ConvAttention2d(i, h, R=0, r=0, yhead=0, xhead=0, ypos=1, xpos=1, bn=1, act=SiLU(True), res=1, group=1, bias=False):
    return Add(
        Seq(
            Add(
                yhead and Attention2d(i, h, r, head=yhead, group=group),
                xhead and Seq(T, Attention2d(i, h, r, head=xhead, group=group), T),
                Conv2d(i, h, 2 * R + 1, padding=R, groups=group, bias=not bn),
                ypos
                and Seq(
                    F(trispace, -1, -3),
                    Conv2d(2, h, 1, bias=False),
                ),
                xpos
                and Seq(
                    F(trispace, -2, -3),
                    Conv2d(2, h, 1, bias=False),
                ),
            ),
            bn and BatchNorm2d(h, 1e-9, track_running_stats=False),
            act,
            Conv2d(h, h, 2 * r + 1, padding=r, groups=group, bias=bias),
        ),
        res and Conv2d(i, h, 2 * r + 1, padding=r, groups=group, bias=False),
    )
```

```python
n, g, h, k, l = 128, 64, 4, 4, 16
m = Seq(
    ConvAttention2d(g, g * k, group=g),
    *(
        Seq(
            ConvAttention2d(g * k, g * k, yhead=h, group=g),
            ConvAttention2d(g * k, g * k, xhead=h, group=g),
        )
        for _ in range(2)
    ),
    ConvAttention2d(g * k, g * l, bias=True),
    Unflatten(-2, [g, l]),
).cuda()
o = torch.optim.Adam(m.parameters())
l = log(
    train(
        lmap(lambda x: [x := torch.rand(n, g, l, device="cuda"), x.argsort(-1)], tqdm(range(10000))),
        m,
        o,
        torch.optim.lr_scheduler.OneCycleLR(o, .01, 10000, pct_start=.1, div_factor=9, final_div_factor=99),
        lambda y, x: [functional.cross_entropy(y.transpose(1, 2), x).mean(), (y.argmax(2)==x).mean([0,2],dtype=torch.float)* 00],
    )
)[-1]
f"{l.mean():.0f}+±{l.std():.0f}" + "".join(f" {l:.0f}" for l in l.quantile(torch.linspace(0, 1, 5)))
```
