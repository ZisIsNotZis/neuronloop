```python
from pandas import read_excel
def Try(f, *_, default=.5):
    try:
        return f(*_)
    except:
        return default
m = {}
for i in 'FE', 'BE1', 'BE2', 'FW', 'HW', 'Test', 'UIUX':
    a = read_excel(r'data/tags.xlsx', i)
    for _,i in a.iterrows():
        _ = i['tag']
        if isinstance(_, str):
            _ = _.strip()
            for j,k in i['Description':][1:].items():
                if isinstance(j, str):
                    m[_,j.strip()] = Try(lambda x:int(x[0]), k)
```

```python
import numpy as np
X = dict(i[::-1]for i in enumerate(set(i[1]for i in m)))
Y = dict(i[::-1]for i in enumerate(set(i[0]for i in m)))
a = np.full((len(X), len(Y)), 0, 'f')
for (i,j),k in m.items():
    a[X[j],Y[i]] = k
```

```python
import torch
from torch.nn import *
from math import prod
class Add(Sequential):
    def forward(s, x):
        return sum((s(x)for s in s[1:]), s[0](x))
class Mul(Sequential):
    def forward(s, x):
        return prod((s(x)for s in s[1:]), start=s[0](x))
class F(Module):
    def __init__(s, f):
        super().__init__()
        s.f = f
    def forward(s, x):
        return s.f(x)
x = torch.tensor(a)
k = 8
m = Sequential(
    Linear(x.size(1), 256),
    Tanh(),
    Linear(256, 2, False),
    Add(
        Sequential(
            Linear(2, k),
            Tanh(),
            Linear(k, x.size(1), False)
        ),
        Sequential(
            Linear(2, k),
            F(torch.sinh),
            Linear(k, x.size(1), False)
        ),
        Sequential(
            Linear(2, k),
            Softmax(1),
            Linear(k, x.size(1), False)
        ),
        Sequential(
            Linear(2, k),
            LogSoftmax(1),
            Linear(k, x.size(1), False)
        ),
        Sequential(
            Mul(
                Sequential(
                    Linear(2, k),
                    Tanh(),
                    Linear(k, k),
                ),
                Linear(2, k),
            ),
            Linear(k, x.size(1), False),
        ),
        Linear(2, x.size(1))
    )
)
o = torch.optim.Adam(m.parameters())
O = torch.optim.lr_scheduler.OneCycleLR(o, .01, 20000, pct_start=.1, div_factor=9, final_div_factor=999)
for _ in range(10):
    for _ in range(2000):
        l = functional.smooth_l1_loss(m(x), x)
        l.backward()
        o.step()
        o.zero_grad()
        O.step()
    print(l)
```

```python
with torch.no_grad():
    A = m[:3](x).numpy()
```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(12,8))
plt.subplots_adjust(0, 0, 1, 1, 0, 0)
plt.scatter(*A.T)
for i,(j,k) in zip(X, A):
    plt.annotate(i, (j, k))
```

```python
import torch
from torch.nn import *
class Add(Sequential):
    def forward(s, x):
        return s[0](x) + s[1](x)
x = torch.tensor(a)
m = Sequential(
    Linear(x.size(1), 512),
    Tanh(),
    Linear(512, 1, False),
    Add(
        Sequential(
            Linear(1, 12),
            Tanh(),
            Linear(12, x.size(1), False)
        ),
        Linear(1, x.size(1))
    )
)
o = torch.optim.Adam(m.parameters())
O = torch.optim.lr_scheduler.OneCycleLR(o, .01, 20000, pct_start=.1, div_factor=9, final_div_factor=999)
for _ in range(10):
    for _ in range(2000):
        l = functional.smooth_l1_loss(m(x), x)
        l.backward()
        o.step()
        o.zero_grad()
        O.step()
    print(l)
```

```python
with torch.no_grad():
    A = m[:3](x).numpy()
```
