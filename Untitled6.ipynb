```python
import numpy as np

A = np.lib.stride_tricks.sliding_window_view(np.memmap("../datbrick/cn1024.H", "h", "r"), 1024)
voc = np.fromfile("../datbrick/cn1024.U1,I", "U1,I")["f0"]
```

```python
from rev import *


def Noise(x, p=0.5):
    global X
    p = torch.randint(x.numel(), [int(x.numel() * p)], device=x.device)
    X = x.ravel().put(p, torch.randint_like(p, len(voc))).view(x.shape)
    return X


def MyPool1d(x):
    return x.unflatten(-1, (-1, 2)).logsumexp(-1)


def Unet(k, m):
    M = []
    for m in m:
        if m:
            M.append(
                Sequential(
                    *(
                        Sequential(
                            Conv1d(k, k, 3, padding=1, bias=False),
                            BatchNorm1d(k),
                            SiLU(True),
                            Conv1d(k, k, 1, bias=False),
                        )
                        for m in range(m)
                    )
                )
            )
        elif m is None:
            M.append(
                RevPair(
                    ConvTranspose1d(k, k := k // 2, 3, 2, 1, 1, bias=False),
                    InvPool1d().T,
                )
            )
        else:
            M.append(
                RevPair(
                    Sequential(
                        Conv1d(k, k, 1, bias=False),
                        F(MyPool1d),
                        Conv1d(k, k := k * 2, 1, bias=False),
                    ),
                    InvPool1d(),
                )
            )
    return Rev(*M)


k = 64
m = Sequential(
    F(Noise),
    Embedding(len(voc), k, scale_grad_by_freq=True),
    F(lambda x: x.transpose(1, 2)),
    Unet(k, [1, 1, 0] * 4 + [1, 1] + [None, 1, 1] * 4),
    Conv1d(k, len(voc), 1),
).cuda()
```

```python
from tqdm.notebook import tqdm
from time import time


def show(o, a, b):
    color = np.full(len(o), "\33[0m", "U5")
    color[(o == a) & (o != b)] = "\33[31m"
    color[(o != a) & (o == b)] = "\33[32m"
    color[(o != a) & (o != b)] = "\33[33m"
    print("".join(np.char.add(color, voc[b])), end="\33[0m\n")


bs = 64
o = torch.optim.Adam(m.parameters())
O = torch.optim.lr_scheduler.OneCycleLR(o, 0.003, len(A) * 10 // (bs * A.shape[1]), pct_start=0.01, div_factor=9, final_div_factor=999)
lastlos = lastlog = time()
for i in tqdm(range(O.last_epoch, O.total_steps + 1)):
    x = torch.tensor(A[np.random.randint(0, len(A), bs, dtype='I')], dtype=torch.long, device="cuda")
    y = m(x)
    l = functional.cross_entropy(y, x)
    l.backward()
    o.step()
    o.zero_grad()
    O.step()
    if time() - lastlos > 30 or i == O.total_steps:
        print(end=f"{l:.2f}:{(y.argmax(1)==x).sum()*100/x.numel():.0f} ")
        lastlos = time()
        # if time() - lastlog > 600 or i == O.total_steps:
        #     x = torch.randint(len(voc), (1, A.shape[1]), device="cuda")
        #     m.eval()
        #     with torch.no_grad():
        #         for p in np.linspace(0.99, 0.01, 100):
        #             m[0].f = lambda x: Noise(x, p)
        #             x = m(x).argmax(1)
        #     m[0].f = Noise
        #     m.train()
        #     print(voc[x[0].cpu()].view(f"U{A.shape[1]}")[0])
        #     lastlog = time()
```
