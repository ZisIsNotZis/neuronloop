```python
import torch
from torch.nn import *
import math
import numpy as np


class F(Module):
    def __init__(s, f, *_):
        super().__init__()
        s.f = f
        s._ = _

    def forward(s, x):
        return s.f(x, *s._)


class Add(Sequential):
    def forward(s, x):
        return sum((s(x) for s in s[1:]), s[0](x))


class Mul(Sequential):
    def forward(s, x):
        return math.prod((s(x) for s in s[1:]), s[0](x))


class Pos(Linear):
    def forward(s, x):
        return x + super().forward(torch.linspace(0, 1, x.size(-2))[:, None])


class TriPos(Linear):
    def __init__(s, _, o, b):
        super().__init__(2, o, b)

    def forward(s, x):
        return x + super().forward(torch.stack((torch.sin(_ := torch.linspace(0, 2 * torch.pi, x.size(-2))), torch.cos(_)), 1))


def Nn(i, h, o=0, bias=False, pos=Identity, bn=Identity, act=ReLU(True)):
    return Sequential(
        Linear(i, h, bn == Identity) if i else Identity(),
        pos(1, h, False),
        bn(h),
        act,
        Linear(h, o, bias) if o else Identity(),
    )


def SwiGLU(i, h, o=0, bias=False, pos=Identity, bn=Identity, act=ReLU(True)):
    return Sequential(
        Mul(
            Sequential(
                (Linear if i else Identity)(i, h, bn == Identity),
                pos(1, h, False),
                bn(h),
                act,
            ),
            Linear(i, h) if i else Identity(),
        ),
        Linear(h, o, bias) if o else Identity(),
    )


class Attention(Module):
    def __init__(s, i, h, o=0, bias=False, nn=[Nn] * 4, bilin=False):
        super().__init__()
        s.q = nn[0](i, h[0], h[4], bilin)
        s.k = nn[1](i, h[1], h[4] if bilin else 0, True)
        s.v = nn[2](i, h[2], h[3])
        s.d = Linear(i, h[3]) if o or i != h[3] else Identity()
        s.o = nn[3](0, h[3], o, bias) if o else Identity()

    def forward(s, x):
        return s.o((s.q(x) @ s.k(x).transpose(-2, -1)).softmax(-1) @ s.v(x) + s.d(x))


class MultiheadAttention(Module):
    def __init__(s, i, h, o, bias=False, head=1, nn=[Nn] * 4, bilin=False):
        super().__init__()
        s.q = Sequential(nn[0](i, head * h[0], head * h[4], bilin), Unflatten(-1, (head, h[4])))
        s.k = Sequential(nn[1](i, head * h[1], head * h[4], bilin), Unflatten(-1, (head, h[4])))
        s.v = Sequential(nn[2](i, head * h[2], head * h[3]), Unflatten(-1, (head, h[3])))
        s.d = Linear(i, head * h[3]) if o or i != head * h[3] else Identity()
        s.o = nn[3](0, head * h[3], o) if o else Identity()

    def forward(s, x):
        return s.o(torch.einsum("...Llh,...lhk->...Lhk", torch.einsum("...Lhk,...lhk->...Llh", s.q(x), s.k(x)).softmax(-2), s.v(x)).flatten(-2) + s.d(x))
```

```python
def Conv(k, n=1, r=1):
    return Sequential(
        *(
            Add(
                Sequential(
                    Conv1d(k, k, r * 2 + 1, padding=r),
                    SiLU(True),
                ),
                Conv1d(k, k, 1, bias=False),
            )
            for n in range(n)
        )
    )

def MyPool1d(x):
    return x.unflatten(-1, (-1, 2)).logsumexp(-1)

def Pool(k, *_):
    return Res(
        AvgPool1d(2),
        Sequential(*_),
        ConvTranspose1d(k, k, 3, 2, 1, 1, bias=False),
    )

def Noise(x, p=.75):
    global X
    X = x.ravel().put(torch.randint(x.numel(), [int(x.numel() * p)], device=x.device), torch.randint(len(voc), [int(x.numel() * p)], device=x.device)).view(x.shape)
    return X 

A = np.lib.stride_tricks.sliding_window_view(np.memmap("../datbrick/py.B", "B", "r"), 8192)
voc = np.fromfile("../datbrick/py.txt", "S1").astype('U1')
k = 64
m = Sequential(
    F(Noise),
    Embedding(len(voc), k, scale_grad_by_freq=True),
    T,
    Conv(k),
    Pool(
        k,
        Conv(k),
        Pool(
            k,
            Conv(k),
            Pool(
                k,
                Conv(k),
                Pool(
                    k,
                    Conv(k),
                    Pool(
                        k,
                        Conv(k),
                        Pool(
                            k,
                            Conv(k),
                            Pool(
                                k,
                                Conv(k),
                                Pool(
                                    k,
                                    Conv(k),
                                    Pool(
                                        k,
                                        Conv(k),
                                        Pool(
                                            k,
                                            Conv(k),
                                            Pool(
                                                k,
                                                Conv(k),
                                            ),
                                            Conv(k),
                                        ),
                                        Conv(k),
                                    ),
                                    Conv(k),
                                ),
                                Conv(k),
                            ),
                            Conv(k),
                        ),
                        Conv(k),
                    ),
                    Conv(k),
                ),
                Conv(k),
            ),
            Conv(k),
        ),
        Conv(k),
    ),
    Conv(k),
    Conv1d(k, len(voc), 1),
).cuda()
```
