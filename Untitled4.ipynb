```python
import numpy as np

A = np.memmap("../datbrick/cn256.B", "B", "r")
```

```python
c = np.zeros(256, "I")
np.add.at(c, A, 1)
c @ np.log2(c) / len(A) - np.log2(len(A))
```

```python
c = np.zeros(65536, "I")
np.add.at(c, np.lib.stride_tricks.sliding_window_view(A, 2).view("H")[:, 0], 1)
c = c[c.astype("?")]
c @ np.log2(c) / len(A) - np.log2(len(A))
```

```python
c = np.zeros(1 << 24, "I")
for i in range(0, len(A), 1048576):
    np.add.at(c, np.pad(np.lib.stride_tricks.sliding_window_view(A, 3)[i : i + 1048576], [[0, 0], [0, 1]]).view("I"), 1)
c = c[c.astype("?")]
c @ np.log2(c) / len(A) - np.log2(len(A))
```

```python
c = np.memmap("tmp.bin", "I", "w+", shape=1 << 32)
np.add.at(c, np.lib.stride_tricks.sliding_window_view(A, 4).view("I")[:, 0], 1)
c = c[c.astype("?")]
c @ np.log2(c) / len(A) - np.log2(len(A))
```

```python
p = 0.2
r = 1 << 13
I = 0
for N in range(1, 100):
    i = np.random.randint(len(A))
    t = np.random.randint(0, 256, A.shape[1], "B")
    _ = np.random.randint(0, A.shape[1], int(A.shape[1] * p))
    t[_] = A[i, _]
    max = (A[i] == t).sum()
    for n in range(0, len(A), r):
        if not (n <= i < n + r) and (A[n : n + r] == t).sum(1).max() >= max:
            I += 1
            break
    print(I * 100 // N, end=" ")
```
