```python
from nn import *
from ipywidgets import Output
from tqdm.notebook import tqdm
import matplotlib.pyplot as plt


class lmap(map):
    def __init__(s, f, _):
        s._ = len(_)

    def __len__(s):
        return s._


def train(x, m, l, a, lr=0.003, pct_start=0.1, div_factor=9, final_div_factor=999, **_):
    o = torch.optim.Adam(m.parameters(), eps=1e-9)
    O = torch.optim.lr_scheduler.OneCycleLR(o, lr, len(x), pct_start=pct_start, div_factor=div_factor, final_div_factor=final_div_factor)

    def train(x):
        y = m(x[0])
        l(y, x[1]).backward()
        o.step()
        o.zero_grad()
        O.step()
        return a(y, x[1])

    log(lmap(train, x), **_)


def log(x, res=500, ref=100):
    res = len(x) // res
    ref = len(x) // ref
    X = [0]
    o = Output()
    display(o)
    try:
        for i, x in enumerate(tqdm(x), 1):
            X[-1] += x
            if not i % res:
                X[-1:] = X[-1][None] / res, 0
                if not i % ref:
                    X = [torch.cat([X.cpu() for X in X[:-1]], 0), 0]
                    plt.plot(X[0])
                    with o:
                        o.clear_output(True)
                        plt.show()
    finally:
        o.clear_output()
        plt.plot(X[0])


class Trispace(Conv1d):
    cache = {}

    def forward(s, x):
        if (_ := s.cache.get(x.size(-1))) is None:
            _ = torch.linspace(0, torch.pi * 2, x.size(-1), device=x.device)
            _ = s.cache[x.size(-1)] = torch.stack([torch.sin(_), torch.cos(_)])
        return super().forward(_)


def squnflatten(x, dim=-1):
    return x.unflatten(dim, [-1, int(x.size(dim) ** 0.5)])


def Bitention(i, o, head=1, group=1):
    i = i // group + 1
    o //= group
    return Seq(
        Unflatten(-2, [group, i - 1]),
        F(functional.pad, [0, 0, 0, 1], value=1),
        Einsum(
            "...ahbc...kc,...jb,...ohjk->...oa",
            Seq(
                Einsum(
                    "...hijk,...kc,...jb,...ia->...ahbc",
                    Param(group, head, i, i, i),
                    Identity(),
                    Identity(),
                    Identity(),
                ),
                Flatten(-2),
                Softmax(-1),
                F(squnflatten),
            ),
            Identity(),
            Identity(),
            Param(group, o, head, i, i),
        ),
        Flatten(-3, -2),
    )


def Attention(i, o, r=0, head=1, group=1):
    i //= group
    o //= group
    return Seq(
        Dot(
            Seq(
                Conv1d(group * i, group * o * head, 2 * r + 1, padding=r, groups=group, bias=False),
                Unflatten(-2, [group, o, head]),
                Flatten(-2),
            ),
            Seq(
                Dot(
                    Seq(
                        Conv1d(group * i, group * head * i, 2 * r + 1, padding=r, groups=group),
                        Unflatten(-2, [group, head, i]),
                        F(torch.permute, [0, 1, 4, 2, 3]),
                    ),
                    Unflatten(-2, [group, 1, i]),
                ),
                Softmax(-1),
                Flatten(-2),
                T,
            ),
        ),
        Flatten(-3, -2),
    )


def ConvAttention(i, o, r=0, attn=0, pos=1, bn=1, act=SiLU(True), res=1, group=1, **_):
    return Add(
        Seq(
            Add(
                attn and Attention(i, o, r, head=attn, group=group),
                Conv1d(i, o, 2 * r + 1, padding=r, groups=group, bias=not bn),
                pos and Trispace(2, o, 1, bias=False),
            ),
            bn and BatchNorm1d(o, 1e-9, track_running_stats=False),
            act,
        ),
        res and Conv1d(i, o, 2 * r + 1, padding=r, groups=group, bias=False),
    )
```

```python
b, g, h, k, l = 8, 2, 4, 4, 16
n, r = 2, 7
train(
    lmap(lambda x: [x := torch.rand(b, 1, l), x.argsort(2)], range(300000)),
    Seq(
        ConvAttention(1, g * k, r, attn=0),
        *[ConvAttention(g * k, g * k, r, attn=h, group=g) for _ in range(n)],
        ConvAttention(g * k, g * k, r, attn=0, group=g),
        ConvAttention(g * k, g, 0, attn=0, bn=0, act=0, res=0, groups=g),
    ),
    lambda y, x: functional.smooth_l1_loss(y, x.float()),
    lambda y, x: (y.round().long() == x).mean((0, 2), dtype=torch.float),
    lr=0.003,
    pct_start=0.1,
    div_factor=9,
    final_div_factor=9999,
)
```
