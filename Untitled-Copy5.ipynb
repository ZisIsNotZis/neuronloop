```python
from nn import *
from ipywidgets import Output
from tqdm.notebook import tqdm
import matplotlib.pyplot as plt


class Trispace(Conv1d):
    def forward(s, x, cache={}):
        if (t := cache.get(x.size(-1))) is None:
            t = torch.linspace(0, torch.pi * 2, x.size(-1), device=x.device)
            t = cache[x.size(-1)] = torch.stack([torch.sin(t), torch.cos(t)])
        return super().forward(t)


def Attention(i, h, bias=False, head=1, sym=1):
    if head == 1:
        if sym == 1:
            return Dot(
                Seq(
                    F(functional.pad, [0, 1], value=1),
                    Dot(Linear(i + 1, i + 1, False), T),
                    Softmax(-1),
                ),
                Linear(i, h, bias),
            )
        return Map(
            lambda w, x, y: w.softmax(-1) @ x + w.transpose(-2, -1).softmax(-1) @ y,
            Seq(
                F(functional.pad, [0, 1], value=1),
                Dot(Linear(i + 1, i + 1, False), T),
            ),
            Linear(i, h, bias),
            Linear(i, h, False),
            unpak=True,
        )
    if sym == 1:
        # ...ai,(hj)i,...bj->...ahb ...ahb,...bj,(hj)k->...ak
        return Seq(
            Dot(
                Seq(
                    F(functional.pad, [0, 1], value=1),
                    Dot(
                        Seq(
                            Linear(i + 1, head * (i + 1), False),
                            Unflatten(-1, [head, -1]),
                        ),
                        F(lambda x: x[..., None, :, :].transpose(-2, -1)),
                    ),
                    Softmax(-1),
                ),
                F(lambda x: x[..., None, :, :]),
            ),
            Flatten(-2),
            Linear(head * i, h, bias),
        )
    return Map(
        lambda w, x, f, g: f((w.softmax(-1) @ x[..., None, :, :]).flatten(-2)) + g((w.transpose(-3, -1).softmax(-1) @ x[..., None, :, :]).flatten(-2)),
        Seq(
            F(functional.pad, [0, 1], value=1),
            Dot(
                Seq(
                    Linear(i + 1, head * (i + 1), False),
                    Unflatten(-1, [head, -1]),
                ),
                F(lambda x: x[..., None, :, :].transpose(-2, -1)),
            ),
        ),
        Identity(),
        Const(Linear(head * i, h, bias)),
        Const(Linear(head * i, h, False)),
        unpak=True,
    )


def ConvAttention(i, h, attn=1, conv=1, pos=0, res=0, bn=BatchNorm1d, act=SiLU(True), Attention=partial(Attention)):
    return Add(
        Seq(
            Add(
                attn and Seq(T, Attention(i, h), T),
                Conv1d(i, h, 2 * conv + 1, padding=conv, bias=not bn),
                pos and Trispace(2, h, 1, bias=False),
            ),
            bn(h, track_running_stats=False),
            act,
        ),
        res and Conv1d(i, h, 1, bias=False),
    )


def GroupAttention(i, h, bias=False, head=1, sym=0, group=1):
    return Seq(
        Unflatten(-1, [group, -1]),
        T2,
        Map(
            lambda w, x, *a: functools.reduce(torch.add, [((w.transpose(-3, -1) if i else w).softmax(-1) @ x).flatten(-2) @ a for i, a in enumerate(a)]),
            Seq(
                F(functional.pad, [0, 1], value=1),
                Dot(
                    Seq(
                        Dot(
                            Identity(),  # gai
                            Param(group, i // group + 1, head * (i // group + 1)),  # gi(hj)
                        ),  # ga(hj)
                        Unflatten(-1, [head, -1]),
                    ),  # gahj
                    F(lambda x: x[..., None, :, :].transpose(-2, -1)),  # g1jb
                ),  # gahb
            ),  # gahb
            F(lambda x: x[..., None, :, :]),  # g1ai / g1bj
            Param(group, head * i // group, h // group),  # g(hj)k
            sym and Param(group, head * i // group, h // group),  # g(hi)k
            unpak=True,
        ),
        T2,
        Flatten(-2),
    )


def GroupAttention(i, h, bias=False, head=1, sym=0, group=1):
    i //= group
    h //= group
    return Seq(
        Unflatten(-2, [group, 1, i]),
        Map(
            list,
            Identity(),
            Seq(
                F(functional.pad, [0, 0, 0, 1], value=1),
                Dot(
                    T,
                    Seq(
                        Flatten(-4, -2),
                        Conv1d(group * (i + 1), group * head * (i + 1), 1, groups=group, bias=False),
                        Unflatten(-2, [group, head, i + 1]),
                    ),
                ),
            ),
        ),
        Add(
            Seq(
                Dot(
                    Identity(),
                    Seq(Softmax(-1), T),
                    zip=True,
                ),
                Flatten(-4, -2),
                Conv1d(group * i, group * h, 1, groups=group, bias=bias),
            ),
            sym
            and Seq(
                Dot(
                    Identity(),
                    Softmax(-2),
                    zip=True,
                ),
                Flatten(-4, -2),
                Conv1d(group * i, group * h, 1, groups=group, bias=False),
            ),
        ),
    )


def GroupBitention(i, h, bias=False, head=1, sym=0, group=1):
    i //= group
    h //= group
    return Seq(
        Unflatten(-2, [group, 1, 1, i]),
        F(functional.pad, [0, 0, 0, 1], value=1),
        Map(
            lambda x: [x[0], x],
            Identity(),
            Dot(
                Seq(
                    Dot(
                        Seq(
                            Flatten(-5, -2),
                            Conv1d(group * (i + 1), group * head * (i + 1) * (i + 1), 1, groups=group, bias=False),
                            Unflatten(-2, [group, head, i + 1, i + 1]),
                            T13,
                        ),
                        Identity(),
                    ),
                    T,
                ),
                Identity(),
            ),
        ),
        Add(
            Seq(
                Dot(
                    Identity(),
                    Seq(
                        Dot(
                            Identity(),
                            Seq(
                                Flatten(-3, -2),
                                Softmax(-2),
                                F(lambda x: x.unflatten(-2, [int(x.size(-2) ** 0.5), -1])),
                            ),
                            zip=True,
                        ),
                        T2,
                    ),
                    zip=True,
                ),
                Flatten(-5, -2),
                Conv1d(group * head * (i + 1) * (i + 1), group * h, 1, groups=group, bias=bias),
            ),
            sym
            and Seq(
                Dot(
                    Identity(),
                    Seq(
                        Dot(
                            Identity(),
                            Seq(
                                T,
                                Flatten(-3, -2),
                                Softmax(-2),
                                F(lambda x: x.unflatten(-2, [int(x.size(-2) ** 0.5), -1])),
                            ),
                            zip=True,
                        ),
                        T2,
                    ),
                    zip=True,
                ),
                Flatten(-5, -2),
                Conv1d(group * head * (i + 1) * (i + 1), group * h, 1, groups=group, bias=False),
            ),
            sym > 1
            and Seq(
                Dot(
                    Identity(),
                    Seq(
                        Dot(
                            Identity(),
                            Seq(
                                T13,
                                Flatten(-3, -2),
                                Softmax(-2),
                                F(lambda x: x.unflatten(-2, [int(x.size(-2) ** 0.5), -1])),
                            ),
                            zip=True,
                        ),
                        T2,
                    ),
                    zip=True,
                ),
                Flatten(-5, -2),
                Conv1d(group * head * (i + 1) * (i + 1), group * h, 1, groups=group, bias=False),
            ),
        ),
    )


def GroupConvAttention(i, h, attn=GroupAttention, conv=1, pos=0, bn=1, act=SiLU(True), res=0, group=1):
    return Add(
        Seq(
            Add(
                attn and attn(i, h, group=group),
                Conv1d(i, h, 2 * conv + 1, padding=conv, bias=not bn, groups=group),
                pos and Trispace(2, h, 1, bias=False),
            ),
            bn and BatchNorm1d(h, track_running_stats=False, eps=1e-9),
            act,
        ),
        res and Conv1d(i, h, 1, bias=False, groups=group),
    )


k, g = 4, 64
```

```python
m = Seq(
    Add(
        Conv1d(1, g * k, 5, padding=2, bias=False),
        Trispace(2, g * k, 1, bias=False),
    ),
    *(GroupConvAttention(g * k, g * k, attn=lambda *_, **__: GroupAttention(*_, **__, head=1), conv=0, pos=1, bn=1, act=SiLU(True), res=1, group=g) for _ in range(2)),
    Conv1d(g * k, g * 16, 5, padding=2, groups=g),
).cuda()
o = torch.optim.Adam(m.parameters(), eps=1e-9)
O = torch.optim.lr_scheduler.OneCycleLR(o, 0.01, 100000, pct_start=0.1, div_factor=9, final_div_factor=99)
l = torch.empty(0, g)
out = Output()
display(out)
for y in tqdm(range(O.last_epoch, O.total_steps)):
    x = torch.rand(64, 16, device="cuda")
    y = m(x[:, None]).unflatten(-2, [g, -1])
    x = x.argsort(1)
    functional.cross_entropy(y.transpose(1, 2), x[:, None].expand(-1, g, -1)).backward()
    o.step()
    o.zero_grad()
    O.step()
    if not (O.last_epoch + 1) % 1000:
        l = torch.cat([l, (y.argmax(2) == x[:, None]).mean((0, 2), dtype=torch.float).cpu()[None]])
        plt.plot(l)
        out.clear_output(True)
        with out:
            plt.show()
```
