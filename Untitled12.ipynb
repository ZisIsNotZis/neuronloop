```python
from nn import *
from ipywidgets import Output
from tqdm.notebook import tqdm
import matplotlib.pyplot as plt


class lmap(map):
    def __init__(s, f, _):
        s._ = len(_)

    def __len__(s):
        return s._


def train(x, m, l, a, lr=0.003, pct_start=0.1, div_factor=9, final_div_factor=999, **_):
    m = m.cuda()
    o = torch.optim.Adam(m.parameters(), eps=1e-9)
    O = torch.optim.lr_scheduler.OneCycleLR(o, lr, len(x), pct_start=pct_start, div_factor=div_factor, final_div_factor=final_div_factor)

    def train(x):
        y = m(x[0])
        l(y, x[1]).backward()
        o.step()
        o.zero_grad()
        O.step()
        return a(y, x[1])

    log(lmap(train, x), **_)
    return m


def log(x, res=1000, ref=2000):
    X = [0]
    o = Output()
    display(o)
    try:
        for i, x in enumerate(tqdm(x), 1):
            X[-1] += x
            if not i % res:
                X[-1:] = X[-1][None] / res, 0
                if not i % ref:
                    X = [torch.cat([X.cpu() for X in X[:-1]], 0), 0]
                    plt.plot(X[0])
                    with o:
                        o.clear_output(True)
                        plt.show()
    finally:
        o.clear_output()
        plt.plot(X[0])


class Trispace(Conv1d):
    cache = {}

    def forward(s, x):
        if (_ := s.cache.get(x.size(-1))) is None:
            _ = torch.linspace(0, torch.pi * 2, x.size(-1), device=x.device)
            _ = s.cache[x.size(-1)] = torch.stack([torch.sin(_), torch.cos(_)])
        return super().forward(_)


def Attention(i, h, head=1, bn=0, group=1, bias=False):
    i //= group
    h //= group
    sym = lambda x, w: torch.cat(
        [
            x @ w.softmax(-1),
            x @ w.mT.softmax(-1),
            x * w.diagonal(0, -2, -1).softmax(-1)[..., None, :],
            x * w.mean(-1, True).mT.softmax(-1),
            x * w.mean(-2, True).softmax(-1),
            *([x * w.mean([-2, -1], True).softmax(-3)] if w.size(-3) > 1 else []),
        ],
        -3,
    )
    return Seq(
        Unflatten(-2, [group, 1, i]),
        Map(
            sym,
            Identity(),
            Seq(
                F(functional.pad, [0, 0, 0, 1], value=1),
                Dot(
                    T,
                    Seq(
                        Flatten(-4, -2),
                        Conv1d(group * (i + 1), group * head * (i + 1), 1, groups=group, bias=False),
                        Unflatten(-2, [group, head, i + 1]),
                    ),
                ),
                bn
                and Seq(
                    Flatten(-4, -3),
                    BatchNorm2d(group * head, 1e-9, track_running_stats=False),
                    Unflatten(-3, [group, head]),
                ),
            ),
            unpak=True,
        ),
        Flatten(-4, -2),
        Conv1d(group * len(sym(torch.empty(1, 1), torch.empty(head, 1, 1))) * i, group * h, 1, groups=group, bias=bias),
    )


def Param(*_):
    return Const(Parameter(init.kaiming_uniform_(torch.empty(_), a=5**0.5)), " ".join(map(str, _)))


def Attention(i, o, head=1, group=1):
    i //= group
    o //= group
    return Seq(
        Unflatten(-2, [group, i]),
        F(functional.pad, [0, 0, 0, 1], value=1),
        Einsum(
            "...gahb,...gjb,...gia,gohij->...goa",
            Seq(
                Einsum(
                    "ghij,...gjb,...gia->...gahb",
                    Param(group, head, i + 1, i + 1),
                    Identity(),
                    Identity(),
                ),
                Softmax(-1),
            ),
            Identity(),
            Identity(),
            Param(group, o, head, i + 1, i + 1),
        ),
        Flatten(-3, -2),
    )


def Attentions(i, o, head=1, group=1, sym="a b"):
    i //= group
    o //= group
    return Seq(
        Unflatten(-2, [group, i]),
        F(functional.pad, [0, 0, 0, 1], value=1),
        Map(
            list,
            Einsum(
                "ghij,...gjb,...gia->...ghab",
                Param(group, head, i + 1, i + 1),
                Identity(),
                Identity(),
            ),
            Identity(),
        ),
        Add(
            "ab" in sym
            and Einsum(
                "...ghab,...gjb,...gia,gohij->...goa",
                F(lambda x: x[0].flatten(-2).softmax(-1).unflatten(-1, x[0].shape[-2:])),
                F(lambda x: x[1]),
                F(lambda x: x[1]),
                Param(group, o, head, i + 1, i + 1),
            ),
            "b" in sym
            and Einsum(
                "...ghab,...gjb,...gia,gohij->...goa",
                F(lambda x: x[0].softmax(-1)),
                F(lambda x: x[1]),
                F(lambda x: x[1]),
                Param(group, o, head, i + 1, i + 1),
            ),
            "a" in sym
            and Einsum(
                "...ghab,...gjb,...gia,gohij->...goa",
                F(lambda x: x[0].softmax(-2)),
                F(lambda x: x[1]),
                F(lambda x: x[1]),
                Param(group, o, head, i + 1, i + 1),
            ),
            "h" in sym
            and Einsum(
                "...ghab,...gjb,...gia,gohij->...goa",
                F(lambda x: x[0].softmax(-3)),
                F(lambda x: x[1]),
                F(lambda x: x[1]),
                Param(group, o, head, i + 1, i + 1),
            ),
            "a-b" in sym
            and Einsum(
                "...ghb,...gjb,...gia,gohij->...goa",
                F(lambda x: x[0].sum(-2).softmax(-1)),
                F(lambda x: x[1]),
                F(lambda x: x[1]),
                Param(group, o, head, i + 1, i + 1),
            ),
            "a-h" in sym
            and Einsum(
                "...ghb,...gjb,...gia,gohij->...goa",
                F(lambda x: x[0].sum(-2).softmax(-2)),
                F(lambda x: x[1]),
                F(lambda x: x[1]),
                Param(group, o, head, i + 1, i + 1),
            ),
            "b-a" in sym
            and Einsum(
                "...gha,...gjb,...gia,gohij->...goa",
                F(lambda x: x[0].sum(-1).softmax(-1)),
                F(lambda x: x[1]),
                F(lambda x: x[1]),
                Param(group, o, head, i + 1, i + 1),
            ),
            "b-h" in sym
            and Einsum(
                "...gha,...gjb,...gia,gohij->...goa",
                F(lambda x: x[0].sum(-1).softmax(-2)),
                F(lambda x: x[1]),
                F(lambda x: x[1]),
                Param(group, o, head, i + 1, i + 1),
            ),
            "i-i" in sym
            and Einsum(
                "...gha,...gja,...gia,gohij->...goa",
                F(lambda x: x[0].diagonal(0, -2, -1).softmax(-1)),
                F(lambda x: x[1]),
                F(lambda x: x[1]),
                Param(group, o, head, i + 1, i + 1),
            ),
            "i-h" in sym
            and Einsum(
                "...gha,...gja,...gia,gohij->...goa",
                F(lambda x: x[0].diagonal(0, -2, -1).softmax(-2)),
                F(lambda x: x[1]),
                F(lambda x: x[1]),
                Param(group, o, head, i + 1, i + 1),
            ),
            "ab-h" in sym
            and Einsum(
                "...gh,...gja,...gia,gohij->...goa",
                F(lambda x: x[0].sum([-2, -1]).softmax(-1)),
                F(lambda x: x[1]),
                F(lambda x: x[1]),
                Param(group, o, head, i + 1, i + 1),
            ),
        ),
        Flatten(-3, -2),
    )


def Bitention(i, o, head=1, group=1):
    i //= group
    o //= group
    return Seq(
        Unflatten(-2, [group, i]),
        F(functional.pad, [0, 0, 0, 1], value=1),
        Einsum(
            "...gahbc,...gkc,...gjb,...gia,gohijk->...goa",
            Seq(
                Einsum(
                    "ghijk,...gkc,...gjb,...gia->...gahbc",
                    Param(group, head, i + 1, i + 1, i + 1),
                    Identity(),
                    Identity(),
                    Identity(),
                ),
                F(lambda x: x.flatten(-2).softmax(-1).unflatten(-1, x.shape[-2:])),
            ),
            Identity(),
            Identity(),
            Identity(),
            Param(group, o, head, i + 1, i + 1, i + 1),
        ),
        Flatten(-3, -2),
    )


def ConvAttention(i, h, attn=1, conv=3, pos=1, bn=1, act=SiLU(True), res=1, group=1, bias=False, **_):
    return Add(
        Seq(
            Add(
                attn and Attention(i, h, **_, group=group),
                Conv1d(i, h, 2 * conv + 1, padding=conv, bias=not bn, groups=group),
                pos and Trispace(2, h, 1, bias=False),
            ),
            bn and BatchNorm1d(h, 1e-9, track_running_stats=False),
            act,
        ),
        res and Conv1d(i, h, 1, groups=group, bias=bias),
    )
```
