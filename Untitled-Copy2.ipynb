```python
from nn import *

A = np.lib.stride_tricks.sliding_window_view(np.memmap("../datbrick/cn.H", "h", "r"), 8192)
unvoc = np.array(list(open("../datbrick/cn.txt", encoding="utf8").read()), "U1")
```

```python
def Conv(k, r=1):
    return Add(
        Seq(
            Conv1d(k, k, r * 2 + 1, padding=r, bias=False),
            BatchNorm1d(k, track_running_stats=False),
            SiLU(True),
        ),
        Conv1d(k, k, 1, bias=False),
    )


def Attn(k, h=1):
    return Res(LiftT(lambda x, _=torch.nn.MultiheadAttention(k, h, batch_first=True): _(x, x, x)[0]))


def Pool(k, *_):
    return Res(
        Conv1d(k, k, 1, bias=False),
        MaxPool1d(2),
        Seq(*_),
        ConvTranspose1d(k, k, 3, 2, 1, 1, bias=False),
    )


def Noise(x, X=len(unvoc), p=0.9):
    return x.ravel().put(torch.randint(x.numel(), [int(x.numel() * p)], device=x.device), torch.randint(X, [int(x.numel() * p)], device=x.device)).view(x.shape)


k = 128
m = Seq(
    F(Noise, p=0.9),
    Peek(F(globals().__setitem__, "X", F.X)),
    Embedding(len(unvoc), k, scale_grad_by_freq=True),
    T,
    Res(
        Conv(k, 5),
        Conv(k),
        Pool(
            k,
            Conv(k),
            Conv(k),
            Pool(
                k,
                Conv(k),
                Conv(k),
                Pool(
                    k,
                    Conv(k),
                    Conv(k),
                    Pool(
                        k,
                        Conv(k),
                        Conv(k),
                        Pool(
                            k,
                            Conv(k),
                            Conv(k),
                            Pool(
                                k,
                                Conv(k),
                                Conv(k),
                                Pool(
                                    k,
                                    Conv(k),
                                    Conv(k),
                                    Pool(
                                        k,
                                        Conv(k),
                                        Conv(k),
                                        Pool(
                                            k,
                                            Conv(k),
                                            Conv(k),
                                            Pool(
                                                k,
                                                Conv(k),
                                                Conv(k),
                                                Pool(
                                                    k,
                                                    Conv(k),
                                                    Conv(k),
                                                ),
                                                Conv(k),
                                                Conv(k),
                                            ),
                                            Conv(k),
                                            Conv(k),
                                        ),
                                        Conv(k),
                                        Conv(k),
                                    ),
                                    Conv(k),
                                    Conv(k),
                                ),
                                Conv(k),
                                Conv(k),
                            ),
                            Conv(k),
                            Conv(k),
                        ),
                        Conv(k),
                        Conv(k),
                    ),
                    Conv(k),
                    Conv(k),
                ),
                Conv(k),
                Conv(k),
            ),
            Conv(k),
            Conv(k),
        ),
        Conv(k),
        Conv(k),
    ),
    Conv1d(k, len(unvoc), 1),
    T,
).cuda()
o = torch.optim.Adam(m.parameters())
O = torch.optim.lr_scheduler.OneCycleLR(o, 5e-4, 100000, pct_start=0.01, div_factor=9, final_div_factor=99)
```

```python
from time import time


def show(o, a, b, **_):
    color = np.full(len(o), "\33[0m", "U5")
    color[(o == a) & (o != b)] = "\33[31m"
    color[(o != a) & (o == b)] = "\33[32m"
    color[(o != a) & (o != b)] = "\33[33m"
    print("".join(np.char.add(color, unvoc[b])), end="\33[0m\n", **_)


with open("log.txt", "w", encoding="utf8") as f:
    lastlos = lastlog = lastsave = time()
    while O.last_epoch < O.total_steps:
        x = torch.tensor(A[np.random.randint(0, len(A), 6)], dtype=torch.long, device="cuda")
        y = m(x)
        l = functional.cross_entropy(y.transpose(1, 2), x)
        l.backward()
        o.step()
        o.zero_grad()
        O.step()
        if time() - lastlos > 30 or O.last_epoch + 1 == O.total_steps:
            print(end=f"{O.last_epoch}:{l:.2f} ", file=f)
            lastlos = time()
            if time() - lastlog > 600 or O.last_epoch + 1 == O.total_steps:
                print(f"{(y.argmax(2)==x).sum()/x.numel()*100:.0f}", file=f)
                show(x[0].cpu(), X[0].cpu(), y[0].argmax(1).cpu(), file=f)
                lastlog = time()
                if time() - lastsave > 1800 or O.last_epoch + 1 == O.total_steps:
                    torch.save((m.state_dict(), o, O), f"b{O.last_epoch}.{l*100:.0f}.pk")
                    lastsave = time()
            f.flush()
```
