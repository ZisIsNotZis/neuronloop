```python
def GroupAttention(i, h, bias=False, head=1, sym=0, group=1):
    i //= group
    h //= group
    return Seq(
        Unflatten(-2, [group, 1, i]),
        Map(
            list,
            Identity(),
            Seq(
                F(functional.pad, [0, 0, 0, 1], value=1),
                Dot(
                    T,
                    Seq(
                        Flatten(-4, -2),
                        Conv1d(group * (i + 1), group * head * (i + 1), 1, groups=group, bias=False),
                        Unflatten(-2, [group, head, i + 1]),
                    ),
                ),
            ),
        ),
        Add(
            Seq(
                Dot(
                    Identity(),
                    Seq(Softmax(-1), T),
                    zip=True,
                ),
                Flatten(-4, -2),
                Conv1d(group * i, group * h, 1, groups=group, bias=bias),
            ),
            sym
            and Seq(
                Dot(
                    Identity(),
                    Softmax(-2),
                    zip=True,
                ),
                Flatten(-4, -2),
                Conv1d(group * i, group * h, 1, groups=group, bias=False),
            ),
        ),
    )


def GroupBitention(i, h, bias=False, head=1, sym=0, group=1):
    i //= group
    h //= group
    return Seq(
        Unflatten(-2, [group, 1, 1, i]),
        F(functional.pad, [0, 0, 0, 1], value=1),
        Map(
            lambda x: [x[0], x],
            Identity(),
            Dot(
                Seq(
                    Dot(
                        Seq(
                            Flatten(-5, -2),
                            Conv1d(group * (i + 1), group * head * (i + 1) * (i + 1), 1, groups=group, bias=False),
                            Unflatten(-2, [group, head, i + 1, i + 1]),
                            T13,
                        ),
                        Identity(),
                    ),
                    T,
                ),
                Identity(),
            ),
        ),
        Add(
            Seq(
                Dot(
                    Identity(),
                    Seq(
                        Dot(
                            Identity(),
                            Seq(
                                Flatten(-3, -2),
                                Softmax(-2),
                                F(lambda x: x.unflatten(-2, [int(x.size(-2) ** 0.5), -1])),
                            ),
                            zip=True,
                        ),
                        T2,
                    ),
                    zip=True,
                ),
                Flatten(-5, -2),
                Conv1d(group * head * (i + 1) * (i + 1), group * h, 1, groups=group, bias=False),
            ),
            sym
            and Seq(
                Dot(
                    Identity(),
                    Seq(
                        Dot(
                            Identity(),
                            Seq(
                                T,
                                Flatten(-3, -2),
                                Softmax(-2),
                                F(lambda x: x.unflatten(-2, [int(x.size(-2) ** 0.5), -1])),
                            ),
                            zip=True,
                        ),
                        T2,
                    ),
                    zip=True,
                ),
                Flatten(-5, -2),
                Conv1d(group * head * (i + 1) * (i + 1), group * h, 1, groups=group, bias=False),
            ),
            sym > 1
            and Seq(
                Dot(
                    Identity(),
                    Seq(
                        Dot(
                            Identity(),
                            Seq(
                                T13,
                                Flatten(-3, -2),
                                Softmax(-2),
                                F(lambda x: x.unflatten(-2, [int(x.size(-2) ** 0.5), -1])),
                            ),
                            zip=True,
                        ),
                        T2,
                    ),
                    zip=True,
                ),
                Flatten(-5, -2),
                Conv1d(group * head * (i + 1) * (i + 1), group * h, 1, groups=group, bias=False),
            ),
        ),
    )
```
