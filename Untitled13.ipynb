```python
from nn import *
from ipywidgets import Output
from tqdm.notebook import tqdm
import matplotlib.pyplot as plt


class lmap(map):
    def __init__(s, f, _):
        s._ = len(_)

    def __len__(s):
        return s._


def train(x, m, l, a, lr=0.003, pct_start=0.1, div_factor=9, final_div_factor=999, **_):
    m = m.cuda()
    o = torch.optim.Adam(m.parameters(), eps=1e-9)
    O = torch.optim.lr_scheduler.OneCycleLR(o, lr, len(x), pct_start=pct_start, div_factor=div_factor, final_div_factor=final_div_factor)

    def train(x):
        y = m(x[0])
        l(y, x[1]).backward()
        o.step()
        o.zero_grad()
        O.step()
        return a(y, x[1])

    log(lmap(train, x), **_)
    return m


def log(x, res=1000, ref=2000):
    X = [0]
    o = Output()
    display(o)
    try:
        for i, x in enumerate(tqdm(x), 1):
            X[-1] += x
            if not i % res:
                X[-1:] = X[-1][None] / res, 0
                if not i % ref:
                    X = [torch.cat([X.cpu() for X in X[:-1]], 0), 0]
                    plt.plot(X[0])
                    with o:
                        o.clear_output(True)
                        plt.show()
    finally:
        o.clear_output()
        plt.plot(X[0])


class Trispace(Conv1d):
    cache = {}

    def forward(s, x):
        if (_ := s.cache.get(x.size(-1))) is None:
            _ = torch.linspace(0, torch.pi * 2, x.size(-1), device=x.device)
            _ = s.cache[x.size(-1)] = torch.stack([torch.sin(_), torch.cos(_)])
        return super().forward(_)


def Attention(i, o, head=1, group=1):
    i //= group
    o //= group
    return Seq(
        Unflatten(-2, [group, i]),
        Einsum(
            "...gahb,...gjb,gohj->...goa",
            Seq(
                F(functional.pad, [0, 0, 0, 1], value=1),
                Einsum(
                    "ghij,...gjb,...gia->...gahb",
                    Param(group, head, i + 1, i + 1),
                    Identity(),
                    Identity(),
                ),
                Softmax(-1),
            ),
            Identity(),
            Param(group, o, head, i),
        ),
        Flatten(-3, -2),
    )


def squnflatten(x, dim=-1):
    return x.unflatten(dim, [-1, int(x.size(dim) ** 0.5)])


def Attentions(i, o, head=1, Head=0, group=1, ab_=1, a_=1, b_=1, a=1, b=1, ab=1):
    i = i // group + 1
    o //= group
    Head = Head or o * i * i
    return Seq(
        Unflatten(-2, [group, i - 1]),
        F(functional.pad, [0, 0, 0, 1], value=1),
        Einsum(
            "...ghab,...gjb,...gia,gohij->...goa" if Head < o * i * i else "...goijab,...gjb,...gia->...goa",
            Seq(
                Einsum(
                    "ghij,...gjb,...gia->...ghab",
                    Param(group, head, i, i),
                    Identity(),
                    Identity(),
                ),
                Flatten(-4, -3),
                Add(
                    ab_
                    and Seq(
                        F(torch.mean, [-2, -1], True),
                        Softmax(-3),
                        Conv2d(group * head, group * Head, 1, groups=group, bias=False),
                    ),
                    b_
                    and Seq(
                        F(torch.mean, -1, True),
                        Softmax(-2),
                        Conv2d(group * head, group * Head, [b_, 1], padding=[b_ // 2, 0], groups=group, bias=False),
                    ),
                    a_
                    and Seq(
                        F(torch.mean, -2, True),
                        Softmax(-1),
                        Conv2d(group * head, group * Head, [1, a_], padding=[0, a_ // 2], groups=group, bias=False),
                    ),
                    b
                    and Seq(
                        Softmax(-1),
                        Conv2d(group * head, group * Head, b, padding=b // 2, groups=group, bias=False),
                    ),
                    a
                    and Seq(
                        Softmax(-2),
                        Conv2d(group * head, group * Head, a, padding=a // 2, groups=group, bias=False),
                    ),
                    ab
                    and Seq(
                        Flatten(-2),
                        Softmax(-1),
                        F(squnflatten),
                        Conv2d(group * head, group * Head, ab, padding=ab // 2, groups=group, bias=False),
                    ),
                ),
                Unflatten(-3, [group, Head] if Head < o * i * i else [group, o, i, i]),
            ),
            Identity(),
            Identity(),
            Head < o * i * i and Param(group, o, Head, i, i),
        ),
        Flatten(-3, -2),
    )


def Bitentions(i, o, head=1, Head=0, group=1, abc_=1, ab_=1, ac_=1, bc_=1, a_=1, a_b=1, a_c=1, b_=1, b_a=1, b_c=1, c_=1, c_a=1, c_b=1, a=1, b=1, c=1, ab=1, ac=1, bc=1, abc=1):
    i = i // group + 1
    o //= group
    Head = Head or o * i * i * i
    return Seq(
        Unflatten(-2, [group, i - 1]),
        F(functional.pad, [0, 0, 0, 1], value=1),
        Einsum(
            "...ghabc,...gkc,...gjb,...gia,gohijk->...goa" if Head < o * i * i * i else "...goijkabc,...gkc,...gjb,...gia->...goa",
            Seq(
                Einsum(
                    "ghijk,...gkc,...gjb,...gia->...ghabc",
                    Param(group, head, i, i, i),
                    Identity(),
                    Identity(),
                    Identity(),
                ),
                Flatten(-5, -4),
                Add(
                    abc_
                    and Seq(
                        F(torch.mean, [-3, -2, -1], True),
                        Softmax(-4),
                        Conv3d(group * head, group * Head, 1, groups=group, bias=False),
                    ),
                    ab_
                    and Seq(
                        F(torch.mean, [-3, -2], True),
                        Softmax(-1),
                        Conv3d(group * head, group * Head, [1, 1, ab_], padding=[0, 0, ab_ // 2], groups=group, bias=False),
                    ),
                    ac_
                    and Seq(
                        F(torch.mean, [-3, -1], True),
                        Softmax(-2),
                        Conv3d(group * head, group * Head, [1, ac_, 1], padding=[0, ac_ // 2, 0], groups=group, bias=False),
                    ),
                    bc_
                    and Seq(
                        F(torch.mean, [-2, -1], True),
                        Softmax(-3),
                        Conv3d(group * head, group * Head, [bc_, 1, 1], padding=[bc_ // 2, 0, 0], groups=group, bias=False),
                    ),
                    (a_ or a_b or a_c)
                    and Seq(
                        F(torch.mean, -3, True),
                        Add(
                            a_
                            and Seq(
                                Flatten(-2),
                                Softmax(-1),
                                F(squnflatten),
                                Conv3d(group * head, group * Head, [1, a_, a_], padding=[0, a_ // 2, a_ // 2], groups=group, bias=False),
                            ),
                            a_b
                            and Seq(
                                Softmax(-2),
                                Conv3d(group * head, group * Head, [1, a_b, a_b], padding=[0, a_b // 2, a_b // 2], groups=group, bias=False),
                            ),
                            a_c
                            and Seq(
                                Softmax(-1),
                                Conv3d(group * head, group * Head, [1, a_c, a_c], padding=[0, a_c // 2, a_c // 2], groups=group, bias=False),
                            ),
                        ),
                    ),
                    (b_ or b_a or b_c)
                    and Seq(
                        F(torch.mean, -2, True),
                        Add(
                            b_
                            and Seq(
                                Flatten(-3),
                                Softmax(-1),
                                F(squnflatten),
                                F(torch.unsqueeze, -2),
                                Conv3d(group * head, group * Head, [b_, 1, b_], padding=[b_ // 2, 0, b_ // 2], groups=group, bias=False),
                            ),
                            b_a
                            and Seq(
                                Softmax(-3),
                                Conv3d(group * head, group * Head, [b_a, 1, b_a], padding=[b_a // 2, 0, b_a // 2], groups=group, bias=False),
                            ),
                            b_c
                            and Seq(
                                Softmax(-1),
                                Conv3d(group * head, group * Head, [b_c, 1, b_c], padding=[b_c // 2, 0, b_c // 2], groups=group, bias=False),
                            ),
                        ),
                    ),
                    (c_ or c_a or c_b)
                    and Seq(
                        F(torch.mean, -1, True),
                        Add(
                            c_
                            and Seq(
                                Flatten(-3, -2),
                                Softmax(-2),
                                F(squnflatten, -2),
                                Conv3d(group * head, group * Head, [c_, c_, 1], padding=[c_ // 2, c_ // 2, 0], groups=group, bias=False),
                            ),
                            c_a
                            and Seq(
                                Softmax(-3),
                                Conv3d(group * head, group * Head, [c_a, c_a, 1], padding=[c_a // 2, c_a // 2, 0], groups=group, bias=False),
                            ),
                            c_b
                            and Seq(
                                Softmax(-2),
                                Conv3d(group * head, group * Head, [c_b, c_b, 1], padding=[c_b // 2, c_b // 2, 0], groups=group, bias=False),
                            ),
                        ),
                    ),
                    a
                    and Seq(
                        Softmax(-3),
                        Conv3d(group * head, group * Head, a, padding=a // 2, groups=group, bias=False),
                    ),
                    b
                    and Seq(
                        Softmax(-2),
                        Conv3d(group * head, group * Head, b, padding=b // 2, groups=group, bias=False),
                    ),
                    c
                    and Seq(
                        Softmax(-1),
                        Conv3d(group * head, group * Head, c, padding=c // 2, groups=group, bias=False),
                    ),
                    ab
                    and Seq(
                        Flatten(-3, -2),
                        Softmax(-2),
                        F(squnflatten, -2),
                        Conv3d(group * head, group * Head, ab, padding=ab // 2, groups=group, bias=False),
                    ),
                    bc
                    and Seq(
                        Flatten(-2),
                        Softmax(-1),
                        F(squnflatten),
                        Conv3d(group * head, group * Head, bc, padding=bc // 2, groups=group, bias=False),
                    ),
                    ac
                    and Seq(
                        T2,
                        Flatten(-2),
                        Softmax(-1),
                        F(squnflatten),
                        Conv3d(group * head, group * Head, ac, padding=ac // 2, groups=group, bias=False),
                        T2,
                    ),
                    abc
                    and Seq(
                        Flatten(-3),
                        Softmax(-1),
                        F(lambda x: x.unflatten(-1, [round(x.size(-1) ** (1 / 3))] * 3)),
                        Conv3d(group * head, group * Head, abc, padding=abc // 2, groups=group, bias=False),
                    ),
                ),
                Unflatten(-4, [group, Head] if Head < o * i * i * i else [group, o, i, i, i]),
            ),
            Identity(),
            Identity(),
            Identity(),
            Head < o * i * i * i and Param(group, o, Head, i, i, i),
        ),
        Flatten(-3, -2),
    )


def ConvAttention(i, h, attn=1, conv=3, pos=1, bn=1, act=SiLU(True), res=1, group=1, bias=False, **_):
    return Add(
        Seq(
            Add(
                attn and Attentions(i, h, **_, group=group),
                Conv1d(i, h, 2 * conv + 1, padding=conv, bias=not bn, groups=group),
                pos and Trispace(2, h, 1, bias=False),
            ),
            bn and BatchNorm1d(h, 1e-9, track_running_stats=False),
            act,
        ),
        res and Conv1d(i, h, 1, groups=group, bias=bias),
    )
```
