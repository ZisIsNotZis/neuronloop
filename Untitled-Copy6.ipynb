```python
from nn import *
from ipywidgets import Output
from tqdm.notebook import tqdm
import matplotlib.pyplot as plt


class Trispace(Conv1d):
    def forward(s, x, cache={}):
        if (t := cache.get(x.size(-1))) is None:
            t = torch.linspace(0, torch.pi * 2, x.size(-1), device=x.device)
            t = cache[x.size(-1)] = torch.stack([torch.sin(t), torch.cos(t)])
        return super().forward(t)


def Attention(i, h, bias=False, head=1, sym=1):
    if head == 1:
        if sym == 1:
            return Dot(
                Seq(
                    F(functional.pad, [0, 1], value=1),
                    Dot(Linear(i + 1, i + 1, False), T),
                    Softmax(-1),
                ),
                Linear(i, h, bias),
            )
        return Map(
            lambda w, x, y: w.softmax(-1) @ x + w.transpose(-2, -1).softmax(-1) @ y,
            Seq(
                F(functional.pad, [0, 1], value=1),
                Dot(Linear(i + 1, i + 1, False), T),
            ),
            Linear(i, h, bias),
            Linear(i, h, False),
            unpak=True,
        )
    if sym == 1:
        # ...ai,(hj)i,...bj->...ahb ...ahb,...bj,(hj)k->...ak
        return Seq(
            Dot(
                Seq(
                    F(functional.pad, [0, 1], value=1),
                    Dot(
                        Seq(
                            Linear(i + 1, head * (i + 1), False),
                            Unflatten(-1, [head, -1]),
                        ),
                        F(lambda x: x[..., None, :, :].transpose(-2, -1)),
                    ),
                    Softmax(-1),
                ),
                F(lambda x: x[..., None, :, :]),
            ),
            Flatten(-2),
            Linear(head * i, h, bias),
        )
    return Map(
        lambda w, x, f, g: f((w.softmax(-1) @ x[..., None, :, :]).flatten(-2)) + g((w.transpose(-3, -1).softmax(-1) @ x[..., None, :, :]).flatten(-2)),
        Seq(
            F(functional.pad, [0, 1], value=1),
            Dot(
                Seq(
                    Linear(i + 1, head * (i + 1), False),
                    Unflatten(-1, [head, -1]),
                ),
                F(lambda x: x[..., None, :, :].transpose(-2, -1)),
            ),
        ),
        Identity(),
        Const(Linear(head * i, h, bias)),
        Const(Linear(head * i, h, False)),
        unpak=True,
    )


def ConvAttention(i, h, attn=1, conv=1, pos=0, res=0, bn=BatchNorm1d, act=SiLU(True), Attention=partial(Attention)):
    return Add(
        Seq(
            Add(
                attn and Seq(T, Attention(i, h), T),
                Conv1d(i, h, 2 * conv + 1, padding=conv, bias=not bn),
                pos and Trispace(2, h, 1, bias=False),
            ),
            bn(h, track_running_stats=False),
            act,
        ),
        res and Conv1d(i, h, 1, bias=False),
    )


def GroupAttention(i, h, bias=False, head=1, sym=0, group=1):
    return Seq(
        Unflatten(-1, [group, -1]),
        T2,
        Map(
            lambda w, x, *a: functools.reduce(torch.add, [((w.transpose(-3, -1) if i else w).softmax(-1) @ x).flatten(-2) @ a for i, a in enumerate(a)]),
            Seq(
                F(functional.pad, [0, 1], value=1),
                Dot(
                    Seq(
                        Dot(
                            Identity(),  # gai
                            Param(group, i // group + 1, head * (i // group + 1)),  # gi(hj)
                        ),  # ga(hj)
                        Unflatten(-1, [head, -1]),
                    ),  # gahj
                    F(lambda x: x[..., None, :, :].transpose(-2, -1)),  # g1jb
                ),  # gahb
            ),  # gahb
            F(lambda x: x[..., None, :, :]),  # g1ai / g1bj
            Param(group, head * i // group, h // group),  # g(hj)k
            sym and Param(group, head * i // group, h // group),  # g(hi)k
            unpak=True,
        ),
        T2,
        Flatten(-2),
    )


def sym(x, w):
    return torch.cat(
        [
            x @ w.softmax(-1),
            x @ w.mT.softmax(-1),
            x * w.diagonal(0,-2,-1).softmax(-1)[..., None, :],
            x * w.mean(-1, True).mT.softmax(-1),
            x * w.mean(-2, True).softmax(-1),
            *([x * w.mean([-2,-1], True).softmax(-3)] if w.size(-3)>3 else []),
        ],
        -3,
    )

def GroupAttention(i, h, bias=False, head=1, sym=sym, group=1, bn=0):
    i //= group
    h //= group
    return Seq(
        Unflatten(-2, [group, 1, i]),
        Map(
            sym,
            Identity(),
            Seq(
                F(functional.pad, [0, 0, 0, 1], value=1),
                Dot(
                    T,
                    Seq(
                        Flatten(-4, -2),
                        Conv1d(group * (i + 1), group * head * (i + 1), 1, groups=group, bias=False),
                        Unflatten(-2, [group, head, i + 1]),
                    ),
                ),
                bn and Seq(
                    Flatten(-4, -3),
                    BatchNorm2d(group * head, track_running_stats=False, eps=1e-9),
                    Unflatten(-3, [group, head]),
                ),
            ),
            unpak=True,
        ),
        Flatten(-4, -2),
        Conv1d(group * len(sym(torch.empty(1, 1), torch.empty(head, 1, 1))) * i, group * h, 1, groups=group, bias=bias),
    )


def sym2(x, w):
    return torch.cat(
        [
            x @ (x @ w.flatten(-3, -2).softmax(-2).unflatten(-2, [-1, w.size(-2)])).transpose(-3, -2),
            x @ (x @ w.transpose(-2, -1).flatten(-3, -2).softmax(-2).unflatten(-2, [-1, w.size(-2)])).transpose(-3, -2),
            x @ (x @ w.transpose(-3, -1).flatten(-3, -2).softmax(-2).unflatten(-2, [-1, w.size(-2)])).transpose(-3, -2),
        ],
        -4,
    )


def GroupBitention(i, h, bias=False, head=1, sym=sym2, group=1):
    i //= group
    h //= group
    return Seq(
        Unflatten(-2, [group, 1, 1, i]),
        F(functional.pad, [0, 0, 0, 1], value=1),
        Map(
            sym,
            Identity(),
            Dot(
                Seq(
                    Dot(
                        Seq(
                            Flatten(-5, -2),
                            Conv1d(group * (i + 1), group * head * (i + 1) * (i + 1), 1, groups=group, bias=False),
                            Unflatten(-2, [group, head, i + 1, i + 1]),
                            T13,
                        ),
                        Identity(),
                    ),
                    T,
                ),
                Identity(),
            ),
        ),
        Flatten(-5, -2),
        Conv1d(group * len(sym(torch.empty(1, 1), torch.empty(head, 1, 1, 1))) * (i + 1) * (i + 1), group * h, 1, groups=group, bias=bias),
    )


def GroupConvAttention(i, h, attn=GroupAttention, conv=3, pos=1, bn=1, act=SiLU(True), res=0, group=1):
    return Add(
        Seq(
            Add(
                attn and attn(i, h, group=group),
                Conv1d(i, h, 2 * conv + 1, padding=conv, bias=not bn, groups=group),
                pos and Trispace(2, h, 1, bias=False),
            ),
            bn and BatchNorm1d(h, track_running_stats=False, eps=1e-9),
            act,
        ),
        res >= 0 and Conv1d(i, h, 2 * res + 1, padding=res, bias=False, groups=group),
    )
```

```python
b, g, h, k, l, r = 16, 1, 1, 32, 256, 3
m = Seq(
    Add(
        Conv1d(1, g * k, 2 * r + 1, padding=r, bias=False),
        Trispace(2, g * k, 1, bias=False),
    ),
    *[GroupConvAttention(g * k, g * k, attn=partial(GroupAttention, head=h, r=r, R=r), conv=r, res=r, group=g) for _ in range(2)],
    Conv1d(g * k, g * l, 2 * r + 1, padding=r, groups=g),
).cuda()
o = torch.optim.Adam(m.parameters(), eps=1e-9)
O = torch.optim.lr_scheduler.OneCycleLR(o, 0.003, 1000000, pct_start=0.01, div_factor=9, final_div_factor=999)
a = [0]
out = Output()
display(out)
for y in tqdm(range(O.last_epoch, O.total_steps)):
    x = torch.rand(b, l, device="cuda")
    y = m(x[:, None]).unflatten(-2, [g, -1])
    x = x.argsort(1)
    functional.cross_entropy(y.transpose(1, 2), x[:, None].expand(-1, g, -1)).backward()
    o.step()
    o.zero_grad()
    O.step()
    a[-1] += (y.argmax(2) == x[:, None]).sum((0, 2))
    if not (O.last_epoch + 1) % 1000:
        a[-1:] = 100/1000/x.numel() * a[-1][None], 0
        if not (O.last_epoch + 1) % 10000:
            a = [torch.cat([a.cpu() for a in a[:-1]]), 0]
            plt.plot(a[0])
            out.clear_output(True)
            with out:
                plt.show()
out.clear_output()
plt.plot(a[0]);
```
