```python
from rev import *


class RevMock(Sequential):
    def forward(s, x):
        x1, x2 = x.chunk(2, 1)
        for s in s:
            if not isinstance(s, RevPair):
                s = RevPair(s)
            x1, x2 = s.g2(x2) + s.f(x1), s.g1(x1)
        return torch.cat([x1, x2], 1)
```
